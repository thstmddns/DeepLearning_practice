```python
import numpy as np
```

```python
a = np.array([0.3, 2.9, 4.0])
exp_a = np.exp(a)
exp_a
```

```python
# 소프트맥스 함수 구현
exp_a_sum = np.sum(exp_a)
exp_a / exp_a_sum
```

```python
# 
def softmax(a):
    exp_a = np.exp(a)
    exp_a_sum = np.sum(exp_a)
    y = exp_a / exp_a_sum

    return y
```
