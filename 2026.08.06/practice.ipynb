```python
# 계단함수 구현(부동소수점)
def step_function(x):
    if x > 0:
        return 1
    else:
        return 0
```

```python
# 계단함수 구현(numpy 입력)
def step_function(x):
    if x > 0:
        return x.astype(int)
```

```python
# sigmoid 함수
def sigmoid_function(x):
    return 1 / (1 + np.exp(-x))
```
