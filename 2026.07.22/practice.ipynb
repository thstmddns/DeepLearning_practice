```python
# AND 조건식
def AND(x1, x2):
    w1, w2, theta = 0.5, 0.5, 0.7
    tmp = w1 * x1 + w2 * x2
    if tmp >= theta:
        return 1
    else:
        return 0
```

```python
# NAND 조건식
import numpy as np
x = np.array([0, 1])
w = np.array([0.5, 0.5])
b = -0.7

print(np.sum(x * w))
print(np.sum(x * w) + b)
```

```python
def NAND(x1, x2):
    w1, w2, theta = -0.5, -0.5, 0.7
    tmp = w1 * x1 + w2 * x2
    if tmp <= theta:
        return 0
    else:
        return 1
```

```python
def NAND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([-0.5, -0.5])
    b = 0.7
    
    tmp = np.sum(x * w) + b
    if tmp <= 0:
        return 0
    else:
        return 1

x1, x2 = -0.5, -0.5
NAND(x1, x2)
```

```python
def OR(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = 0.2

    tmp = np.sum(x * w) + b
    if tmp <= 0:
        return 0
    else:
        return 1
```

```python
def XOR(x1, x2):
    s1 = OR(x1, x2)
    s2 = NAND(x1, x2)
    result = AND(s1, s2)
    return result
```
